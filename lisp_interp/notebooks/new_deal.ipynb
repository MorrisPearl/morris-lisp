## The name, the dates, the pricing speed and the Treasury curve

The pricing speed is a type (PSA, CPR or SMM) and a row for each piece of it, which goes in a straight line from *from* to *to*, and is *to* in the last of its *months* (0 months, for the last piece, is the rest of the time). The Treasury curve is a row for each point, its years and its yield in percent: a spread is over it, at the average life of a tranche.

## The collateral

A row for each piece: its balance, its wac and its pass-through coupon in percent, its wam and its age in months. **Show the cash flows** adds the pieces up at the pricing speed; **Update the deal** gives the deal what is in the grid.

```scheme
(collateral deal)
```

## The tranches

Click a tranche to divide it, or to change its split. Click *view* in the *cash flows* column for the cash flows of a tranche. Type a price, a yield or a spread in its cell (double-click it, or click it and type: a `+` or `?` is typed after the double-click) and the others are worked out, with the duration, the dv01 and the money: a price can be `+`, for what its children are worth together, or `?`, for what makes it and its siblings worth their parent. The delay, in days, is typed too. A cleared cell takes the price away.

```scheme
(bonds deal)
```

When the deal is built, run it at a speed (a prepayment object), and look at it as usual. The index is for the floaters, and `deal-average-life` is the average life of a bond, in years.

```scheme
(deal-run! deal (prepayment-speed 'psa 200) :index 0.04)
(deal-check deal)
(display-table (deal-outline deal) '(("balance" ",.2f")))
```
