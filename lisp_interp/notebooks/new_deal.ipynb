# A deal, in three views

A deal is an object. It has the data of its collateral, its tranches, and a few other things: its dates and its pricing speed. Each view of it is a cell, and they can all be shown at once: what is changed in one is shown in the others. See "The deal object, and its views in a notebook" in `lisp_cmo_manual.md`.

```scheme
(define deal (new-deal "Demo 2026-1" :first-payment (date 2026 11 25) :closing-date (date 2026 11 20)))
```

## The name, the dates and the pricing speed

The pricing speed is a type (PSA, CPR or SMM) and a row for each piece of it, which goes in a straight line from *from* to *to*, and is *to* in the last of its *months* (0 months, for the last piece, is the rest of the time).

```scheme
(info deal)
```

## The collateral

A row for each piece: its balance, its wac and its pass-through coupon in percent, its wam and its age in months. **Show the cash flows** adds the pieces up at the pricing speed; **Update the deal** gives the deal what is in the grid.

```scheme
(collateral deal)
```

## The tranches

Click a tranche to divide it, or to change its split. Click *view* in the last column for the cash flows of a tranche.

```scheme
(bonds deal)
```

When the deal is built, run it at a speed (a prepayment object), and look at it as usual. The index is for the floaters, and `deal-average-life` is the average life of a bond, in years.

```scheme
(deal-run! deal (prepayment-speed 'psa 200) :index 0.04)
(deal-check deal)
(display-table (deal-outline deal) '(("balance" ",.2f")))
```
