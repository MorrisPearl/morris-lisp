# Starting a deal

`new-deal` shows a grid for the collateral, the pricing speed, and two buttons. Type a row for each piece of collateral (its balance, wac and pass-through coupon in percent, wam and age in months), choose the speed (PSA, CPR or SMM; one row for a single speed, or a row for each piece of a speed that changes), press **Show the cash flows** to see the pieces added up, and **Create the deal**: the deal builder takes the place of the grids, and the deal is the variable `deal`.

See "Starting a deal in a notebook" in `lisp_cmo_manual.md`.

```scheme
(define start (new-deal "Demo 2026-1" (date 2026 11 25) :variable 'deal))
start
```

When the deal is divided into tranches, run it and look at it as usual. The index is for the floaters, and `deal-average-life` is the average life of a bond, in years.

```scheme
(deal-run! deal :index 0.04)
(deal-check deal)
(display-table (deal-outline deal) '(("balance" ",.2f")))
```
