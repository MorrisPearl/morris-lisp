# A shock to the AI companies

What happens to the rest of the economy if the companies building AI stop spending: (a) a moratorium on new data centers,
(b) a solvency crisis, or (c) demand for AI more elastic than thought? The four-household model of `four_kinds_of_households.ipynb`,
with an AI bust in four parts: AI investment falls, AI shares fall, lenders grow fearful, and AI gets cheaper for the firms that use it.
Every size is an assumption: the section *A shock to the AI companies* of `doc/lisp_macro_models_manual.md` says what each one is,
which results come from them, and what this model can't say.

## 1. How big the boom is

Business investment in computers, most of it servers for data centers, as a percent of GDP. It was about half a percent before the boom.

```scheme
(define boom (ai-investment-data creds))
(plot-chart (list (list "computer investment" (table-column boom "date") (table-column boom "percent of GDP")))
            :title "Investment in computers and peripheral equipment" :y-label "percent of GDP")
```

## 2. The three scenarios

Each is a set of the model's `ai_` parameters.

```scheme
(ai-bust-scenarios)
```

```scheme
(define busts (ai-bust-models))
(plot-ai-busts busts)
```

```scheme
(display-table (compare-ai-busts busts))
```

## 3. Which part does what

The solvency crisis, and each of its parts alone. The model is linear, so the parts add up to the whole: the credit spread is what makes a recession.

```scheme
(display-table (compare-ai-busts (ai-bust-parts (second (second (ai-bust-scenarios))))))
```

## 4. Who bears it

The solvency crisis by group of households: the top 1% owns half the stocks; the bottom half's income moves most with jobs.

```scheme
(plot-ai-bust (second (second busts)))
```

## 5. Changing an assumption

Any parameter can be changed in every scenario: here, more lender fear in the crisis, and a Fed slow to cut.

```scheme
(define crisis (second (second (ai-bust-scenarios))))
(display-table (compare-ai-busts (ai-bust-models :scenarios (list (list "as assumed" crisis)
                                                               (list "2 points of credit stress" (append crisis '((ai_credit_stress . 2))))
                                                               (list "Fed slow to cut" (append crisis '((fed_smoothing . 0.95))))))
                                 :quarters '(1 4 20)))
```
