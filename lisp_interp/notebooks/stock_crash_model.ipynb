# What a stock market crash does to the US economy

A small macroeconomic model with a *financial accelerator*, run with Dynare. Every function here is explained in
`doc/lisp_macro_models_manual.md`, which also says what the model leaves out. Responses are by the quarter:
quarter 1 is the quarter of the crash. "The stock market" here is firms' **net worth**: what they own less what they owe.

## 1. A 20% fall in the stock market

`crash-model` runs Dynare (a second or two). `stock-crash-responses` is a table of what happens when firms' net worth is 20% below
normal in the first quarter; GDP, consumption, investment and net worth are percent from normal, inflation, the Fed's rate and the
credit spread are points a year.

```scheme
(define base (crash-model))
(display-table (stock-crash-responses base :fall 20))
```

```scheme
(plot-stock-crash base :fall 20)
```

## 2. What the accelerator adds

`accelerator` is how much the credit spread rises when firms borrow more compared with their own net worth. At 0 the financial
channel is shut: the same crash, with no accelerator.

```scheme
(define without (crash-model :parameters '((accelerator . 0))))
(plot-stock-crash (list (list "with the accelerator" base) (list "without" without)) :fall 20)
```

## 3. Scenarios: change a parameter, compare

`leverage` is how much firms borrow (capital compared with their own net worth); `fed_inflation_response` and `fed_smoothing` describe
the Fed's rule. Ten quarters this time.

```scheme
(define cautious (crash-model :parameters '((leverage . 1.5)) :quarters 10))
(define leveraged (crash-model :parameters '((leverage . 3)) :quarters 10))
(plot-stock-crash (list (list "leverage 1.5" cautious) (list "leverage 3" leveraged)) :fall 20)
```

```scheme
(define quick-fed (crash-model :parameters '((fed_smoothing . 0.2)) :quarters 10))
(define slow-fed (crash-model :parameters '((fed_smoothing . 0.9)) :quarters 10))
(plot-stock-crash (list (list "Fed moves fast" quick-fed) (list "Fed moves slowly" slow-fed)) :fall 20)
```

## 4. Other kinds of surprise

A crash can also come as lenders' fears (the credit spread jumps) or as a loss of confidence (demand). These are tables of one typical surprise.

```scheme
(display-table (crash-model-responses base "lender_fear_shock"))
(display-table (crash-model-responses base "demand_shock"))
```

## 5. The parameters estimated from US data, 1985 to 2008

About a quarter of an hour; `:draws 1000` gives a quicker first look. The data end before the Fed's rate hit zero, which the model can't capture.

```scheme
(define estimate (crash-model-estimate creds))
(show-crash-estimates estimate)
```

```scheme
(plot-stock-crash estimate :fall 20)
```

## 6. A scenario that starts from the estimate

```scheme
(define estimated-without (crash-model :from estimate :parameters '((accelerator . 0))))
(plot-stock-crash (list (list "as estimated" estimate) (list "no accelerator" estimated-without)) :fall 20)
```
