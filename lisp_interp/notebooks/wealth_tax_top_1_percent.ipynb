# A wealth tax on the top 1%

The four-household model of `four_kinds_of_households.ipynb`, with a government that taxes the wealth of the top 1%. This is an
experiment under assumptions: the section *A wealth tax on the top 1%* of `doc/lisp_macro_models_manual.md` says what each one is, which
results come from an assumption rather than the data, and what the model leaves out.

## 1. A tax of 1% a year

The tax starts in quarter 1 and fades very slowly, so it is close to permanent. Forty quarters is ten years. The revenue is paid out equally
to every household, and a fifth of the tax is avoided.

```scheme
(define tax (types-model :quarters 40))
(plot-wealth-tax tax :rate 1)
```

```scheme
(display-table (wealth-tax-responses tax :rate 1) :max-rows 12)
```

## 2. How much does each assumption matter?

`compare-wealth-taxes` puts the same tax under different assumptions side by side (a row for each assumption and each of the quarters 4, 20
and 40). Compare the effect on investment across the assumptions about how much of the tax comes out of firms' net worth.

```scheme
(define scenarios
  (list (list "as assumed" tax)
        (list "none avoided" (types-model :quarters 40 :parameters '((tax_avoidance . 0))))
        (list "half avoided" (types-model :quarters 40 :parameters '((tax_avoidance . 0.5))))
        (list "nothing from firms" (types-model :quarters 40 :parameters '((tax_from_firm_equity . 0))))
        (list "twice from firms" (types-model :quarters 40 :parameters '((tax_from_firm_equity . 0.5))))))
(display-table (compare-wealth-taxes scenarios) :max-rows 20)
```

## 3. What the government does with the revenue

The calibration pays it out equally to every household. Compare paying it all to the bottom half, spending it on goods and services,
and using it to retire debt (which, in this model, changes nothing).

```scheme
(define nobody '((revenue_to_bottom50 . 0) (revenue_to_next40 . 0) (revenue_to_next9 . 0) (revenue_to_top1 . 0)))
(define uses
  (list (list "paid out equally" tax)
        (list "all to the bottom half"
              (types-model :quarters 40 :parameters '((revenue_to_bottom50 . 1) (revenue_to_next40 . 0) (revenue_to_next9 . 0) (revenue_to_top1 . 0))))
        (list "spent on goods and services"
              (types-model :quarters 40 :parameters (append nobody '((revenue_to_purchases . 1)))))
        (list "retires debt" (types-model :quarters 40 :parameters nobody))))
(display-table (compare-wealth-taxes uses) :max-rows 20)
```

## 4. A bigger or smaller tax

The model is linear: a tax of 2% a year does twice what a tax of 1% does (and, being linear, it is only reliable for taxes of a few percent).

```scheme
(display-table (compare-wealth-taxes (list (list "2% a year" tax)) :rate 2))
```
