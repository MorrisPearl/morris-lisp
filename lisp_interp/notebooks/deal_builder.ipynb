# Building a deal by clicking

A deal begins as one tranche, the collateral. `deal-builder` shows a grid of the tranches of the deal.

- **Click a tranche** to divide it: choose how in the panel below (a tab for each kind of split), fill in the fields, press **Calculate the sizes** to look at the tranches, and **Add the tranches to the deal** to add them. The new tranches appear under the one divided, so the grid is the tree of the deal, depth first.
- **Click a tranche that is divided** to see its split, with the inputs it was made with. Change them, press **Calculate the sizes** to see what everything under it would become, and **Change the split** to make it so: the tranches of the split, and those under them, are worked out again.
- **Take away this split, and all under it** removes the split and every tranche below it (it asks first).

See "Building a deal by clicking" in `lisp_cmo_manual.md`.

```scheme
(define deal (new-deal "Demo 2026-1" :collateral '((100000000 6.5 6.5 360)) :speed (prepayment-speed 'psa 150)))
(deal-builder deal)
```

When the deal is built, run it and look at it as usual. The index is for the floaters.

```scheme
(deal-run! deal :index 0.04)
(deal-check deal)
(display-table (deal-outline deal) '(("balance" ",.2f")))
```
